```python
experiments['assignment_date'] = pd.to_datetime(experiments['assignment_date'])
users['signup_date'] = pd.to_datetime(users['signup_date'])

check = experiments.merge(
    users[['user_id', 'signup_date']],
    on='user_id',
    how='left'
)

(check['assignment_date'] < check['signup_date']).sum()
```

```python
pd.crosstab(
    experiments['experiment_group'],
    experiments['variant_version']
)
```

```python
invalid_assignments = check[
    check['assignment_date'] < check['signup_date']
].sort_values('assignment_date')

invalid_assignments.head(10)
```

```python
invalid_assignments[
    ['user_id', 'signup_date', 'assignment_date',
     'experiment_group', 'variant_version']
]
```

```python
invalid_assignments['days_before_signup'] = (
    invalid_assignments['signup_date']
    - invalid_assignments['assignment_date']
).dt.days

invalid_assignments['days_before_signup'].value_counts().sort_index()
```
